## Spark Interview Questions

1.What is SparkContext?

Answer: <br>
SparkContext is the entry point to Spark's core functionality. It represents a connetion to a spark cluster and is responsible for RDD creation, task scheduling, and resource coordination.

2.Why is SparkContext needed?

Answer: <br>
Without SparkContext, Spark cannot: <br>

Connect to the cluster <br>

Create RDDs <br>

Submit jobs <br>

Access Spark configuration <br>

It acts as the driver's gateway to the cluster.

3.Difference between SparkContext and SparkSession?

Answer: <br>

SparkContext -> low-level API (RDD) <br>

SparkSession -> unified entry point for DataFrame, SQL, Streaming <br>

In Spark 2.x+, SparkSession internally contains SparkContext

4.What is an RDD in Spark?

Answer: <br>
An RDD (Resilient Distributed Dataset) is Spark's core data structure. It is an immutable, distributed collection of objects that can be processed in parallel and is fault-tolerant through lineage.

5.What is lazy evaluation in RDD? <br>
Answer: <br>
RDD transformation are not executed immediately. They are recoreded and executed only when an action is called. This allow Spark to optimize execution.
