####Azure Cloud
1.Create Resource Group<br>
2.Create Virtual Network within above Resource Group name<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Create Subnets (1:Public Subnet - IPv4(10.0.1.0/24 & Size /25) 2:Private Subnet - IPv4(10.0.1.128/24 & Size /25))<br>
3.Create Databricks Workspace within above Resource Group name<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Attach the Public and Private subnets with Subnet name & Subnet CIDR range<br>
4.Create Azure Data Lake Storage Gen2 within above Resource Group name (Eg: adlswithdatabricks001)<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Create container within storage (Eg: root)<br>
&nbsp;&nbsp;&nbsp;&nbsp;b.Create folder/directory within root container (Eg: metastore)<br>
5.Create Access Connector for Azure Databricks within above Resource Group name (Eg: adb-uc-connector)<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Give permission to this connector within above created ADLS Gen2 account<br>
&nbsp;&nbsp;&nbsp;&nbsp;b.Within ADLS Gen2 created account go to Access Control (IAM) add role as Storage Blob Data Contributor and select managed identity and then select just created Access Connector<br>
6.Open URL: accounts.azuredatabricks.net<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Click on catalog tab<br>
&nbsp;&nbsp;&nbsp;&nbsp;b.Click on Create Metastore<br>
&nbsp;&nbsp;&nbsp;&nbsp;c.Give me Metastore name (Eg: azure-central-india) and then attach the ADLS Gen2 path ie. 'root@adlswithdatabricks001.dfs.core.windows.net/metastore'<BR>
&nbsp;&nbsp;&nbsp;&nbsp;d.Add access connector id (Go to access connector for databricks and copy the Resource ID and Paste) and click on create

#### Note:<br>
1.If you don't configure an external location at CATALOG level then managed catalog,schema,table gets created and stored at METASTORE location ie. 'root@adlswithdatabricks001.dfs.core.windows.net/metastore'<br>
2.But, If you configure an external location at CATALOG level then managed schema,table gets created and stored at CATALOG external location<br>
3.If you don't configure an external location at SCHEMA level then managed schema, table gets created and stored at CATALOG location ie. 'data@adlswithdatabricks001.dfs.core.windows.net/catalog'<br>
4.But, If you configure an external location at SCHEMA level then managed schema, table gets created and stored at SCHMEA external location<br>
5.How storing data flow: Table to Schema; Schema to Catalog; Catalog to Metastore - Wherever top level location found while creating an object there the data gets created and stored.

####Configuration an Extranal location to Catalog level
1.Open Databricks Workspace and go to catalog tab.<br>
2.Click on Storage credential tab which is displayed on main panel.<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Copy Resource Id from Access Connector for Azure Databricks and paste.<br>
&nbsp;&nbsp;&nbsp;&nbsp;b.Give good Storage credential name and click on save.<br>
3.Create one more container (Eg: data) and under it one folder (Eg: catalog) within above created ADLS Gens account.<br>
4.Go to Databricks Workspace > Catalog > External Location<br>
&nbsp;&nbsp;&nbsp;&nbsp;a.Provide Catalog External Location<br>
&nbsp;&nbsp;&nbsp;&nbsp;b.Provide Storage credential name which you created earlier
