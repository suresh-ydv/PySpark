```python
# CREATE MANAGED TABLE
CREATE TABLE DEV.BRONZE.SALE_MANAGED(
    invoice_id STRING,
    cust_id STRING,
    product_code STRING
);
```

```python
# CREATE EXTERNAL TABLE
CREATE TABLE DEV.BRONZE.SALE_EXTERNAL(
    invoice_id STRING,
    cust_id STRING,
    product_code STRING
)
LOCATION '/mnt/bronze/sale'
PARTITIONED BY (product_code);
```

```python
# Table will be stored in the default METASTORE location
DESCRIBE EXTENDED DEV.BRONZE.SALE_MANAGED;
```

```python
# On drop data and metadata will be deleted
DROP TABLE DEV.BRONZE.SALE_MANAGED;
```

```python
# Table will be stored in the specified location
DESCRIBE EXTENDED DEV.BRONZE.SALE_EXTERNAL;
# On drop only metadata will be deleted
DROP TABLE DEV.BRONZE.SALE_EXTERNAL;
```

```python
# To see the dropped table
USE CATALOG DEV;

SHOW TABLES DROPPED IN BRONZE;
```

```python
# To undrop the table
UNDROP TABLE DEV.BRONZE.SALE_MANAGED;
```
